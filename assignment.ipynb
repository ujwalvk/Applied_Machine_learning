```python
#8
sal = np.array([25000,32000,28000,45000,50000,38000,42000,30000])

print('Total salary: \n', sal.sum())
print('Average salary: \n', sal.mean())
print('Total salary: \n', sal.sum())
print('Highest salary: \n', sal.max())
print('Lowest salary: \n', sal.min())
print('Median salary: \n', np.median(sal))
print('Standard deviation: \n', sal.std())
```

```python
#9
marks = np.array([45,78,89,32,67,90,55,76,41,84])
print("Score greater than 75 : ",marks[marks>75])
print("\n Score between 50 and 80: ", marks[(marks>50) & (marks<80)])
print('\n Number of students who scored less than 50: ',len(marks[marks<50]))
p=(len(marks[marks>60])/len(marks))*100
print('\n Percentage of students who scored 60 or above :', p)
```

```python
marks = np.array([
    [80,75, 90],
    [65,70,72],
    [88,92,85],
    [55,60,58]
])

print("Total of Student 1 : ", marks[0].sum())
print("Total of Student 2 : ", marks[1].sum())
print("Total of Student 3 : ", marks[2].sum())
print("Total of Student 4 : ", marks[3].sum())

print("Average marks  of Student 1 : ", marks[0].mean())
print("Average marks  of Student 2 : ", marks[1].mean())
print("Average marks  of Student 3 : ", marks[2].mean())
print("Average marks  of Student 4 : ", marks[3].mean())

print("Average in subject 1 : ", np.mean(marks, axis =0))
```
