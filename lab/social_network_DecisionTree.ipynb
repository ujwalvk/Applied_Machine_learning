```python
df#independent variables
x = df.iloc[:, 2:-1].values
x
```

```python
y=df.iloc[:,-1].values
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
#standardises (normalises)
#ensures that columns with large numbers (like Income) do not dominate columns with small numbers (like Age) during model training
from sklearn.preprocessing import StandardScaler
sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
```

```python
X_test
```

```python
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
#classifier = DecisionTreeClassifier(criterion = 'gini', random_state = 0)
classifier.fit(X_train, y_train)
```

```python
#predicting test set results
y_pred = classifier.predict(X_test)
y_pred
```

```python
y_test
```

```python
#confusion Matrix
from sklearn.metrics import confusion_matrix# confusion_matrix is not a class
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
#62, 29 predticted as correct
#6,3 incorrect classification
"""
  0  1
0[62,6]
  TN FP
1[3,29]
  FN TP
"""
```

```python
accuracy = (cm[0][0]+cm[1][1])/(cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1])
accuracy
```

```python
from sklearn  import metrics
prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy: ",prediction*100,'%')
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
#predict for a new individual with age =29 and salary = 80000
new_individual = np.array([[29, 80000]])

#scale new individual features
new_individual_scaled=sc_X.transform(new_individual)
#predict using trained classifier
prediction = classifier.predict(new_individual_scaled)
if prediction[0] == 1:
    print("The person will  buy the product")
else:
    print("The person will not buy the product")
```
